```python
import json
from datetime import datetime

# Load the JSON data
with open("../../../data/cpi_site_data/mhhi_site_data/data_download_script/daily_hhi_cpi.json", "r") as file:
    data = json.load(file)

# Function to convert dates to MM-DD-YYYY format
def convert_date_format(date_str):
    try:
        # Check if the date is in MM-DD-YYYY format (with dashes)
        return datetime.strptime(date_str, "%m-%d-%Y").strftime("%m-%d-%Y")
    except ValueError:
        # If not, try the MM/DD/YYYY format (with slashes)
        return datetime.strptime(date_str, "%m/%d/%Y").strftime("%m-%d-%Y")

# Update the date format in the JSON data
for entry in data:
    entry["date"] = convert_date_format(entry["date"])

# Save the updated JSON data back to the file
with open("../../../data/cpi_site_data/mhhi_site_data/data_download_script/daily_hhi_cpi.json", "w") as file:
    json.dump(data, file, indent=4)

print("Date format has been standardized to MM-DD-YYYY.")
```
