```python
import csv
import json

# Specify the input CSV file and output JSON file paths
csv_file_path = '../../../data/cpi/line_graph_data/daily_hhi_cpi.csv'
json_file_path = '../../../data/cpi_site_data/mhhi_site_data/data_download_script/daily_hhi_cpi.json'

# Initialize an empty list to store the rows as dictionaries
data_list = []

# Open the CSV file for reading
with open(csv_file_path, mode='r', encoding='utf-8') as csv_file:
    # Create a CSV reader object
    csv_reader = csv.DictReader(csv_file)
    
    # Loop through each row in the CSV file and add it to the list
    for row in csv_reader:
        data_list.append(row)

# Write the list of dictionaries to the JSON file
with open(json_file_path, mode='w', encoding='utf-8') as json_file:
    json.dump(data_list, json_file, indent=4)

print(f"CSV file has been converted to JSON format and saved as {json_file_path}")
```
