```python
import pandas as pd
from datetime import timedelta

# Load the CSV data
data = pd.read_csv('../../../data/cpi_site_data/mhhi_site_data/data_download_script/optimism_data.csv')

# Convert the 'day' column to datetime for sorting and processing
data['day'] = pd.to_datetime(data['day'])

# Sort the data by day
data.sort_values('day', inplace=True)

# Dictionary to store the voting power of delegates from the last processed day
last_day_data = {}

# Get the range of dates from the first to the last day in the data
all_dates = pd.date_range(start=data['day'].min(), end=data['day'].max())

# Iterate through each date in the full range
for day in all_dates:
    # Filter data for the current day
    current_day_data = data[data['day'].dt.date == day.date()]
    
    # Dictionary to store the voting power for the current day
    current_day_dict = {}

    # Update the current day's data based on the previous day's data
    for _, row in current_day_data.iterrows():
        delegate = row['delegate']
        voting_power = row['voting_power']
        current_day_dict[delegate] = voting_power
    
    # If it's not the first day, carry forward the previous day's data
    if last_day_data:
        for delegate, voting_power in last_day_data.items():
            if delegate not in current_day_dict:
                current_day_dict[delegate] = voting_power

    # Update the last_day_data for the next iteration
    last_day_data = current_day_dict

    # Convert the dictionary to a DataFrame
    output_df = pd.DataFrame(list(current_day_dict.items()), columns=['delegate', 'voting_power'])
    
    # Save to CSV, using the day as the filename
    output_df.to_csv(f'{day.date()}.csv', index=False)

print("CSV files created for each day in the range.")
```

```python
import csv
import json

# Define the input and output file paths
csv_file_path = '../../../data/cpi/optimism/output_hhi_cpi.csv'
json_file_path = '../../../cpi_github_repo/cpi/public/output_hhi_cpi.json'

# Read the CSV file and convert to JSON format
data = []

# Open the CSV file and read its contents
with open(csv_file_path, mode='r') as csv_file:
    csv_reader = csv.DictReader(csv_file)
    
    # Iterate over each row and convert it to a dictionary
    for row in csv_reader:
        data.append(row)

# Write the JSON data to a file
with open(json_file_path, mode='w') as json_file:
    json.dump(data, json_file, indent=4)

print(f"CSV file '{csv_file_path}' has been successfully converted to JSON file '{json_file_path}'.")
```
