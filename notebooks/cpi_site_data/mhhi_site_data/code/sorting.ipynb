```python
import pandas as pd

# Read the CSV file
file_path = '../../../../data/cpi_site_data/mhhi_site_data/data_download_script/uniswap_data.csv'
df = pd.read_csv(file_path)

# Sort the data in descending order based on the 'th_vp' column
df_sorted = df.sort_values(by='th_vp', ascending=False)

# Save the sorted data to a new CSV file (optional)
df_sorted.to_csv('uniswap_data_sorted.csv', index=False)

# Display the sorted DataFrame (optional)
print(df_sorted)
```
